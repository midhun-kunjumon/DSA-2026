```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('/content/iris.csv')
df.head()
```

```python
columns = df.columns
print("The columns in the dataset are:")
print(columns)
```

```python
mean_values = df.mean(numeric_only=True)
print("Mean of each column:")
print(mean_values)
```

```python
null_values = df.isnull().sum()
print("Number of null values in each column:")
print(null_values)
```

```python
avg_petal_length = df.groupby('Classification')['PL'].mean()
avg_petal_length.plot(kind='bar', color=['pink', 'skyblue', 'lightgreen'])
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (PL)')
plt.show()
```

```python
plt.scatter(df['SL'], df['SW'], color='purple')
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (SL)')
plt.ylabel('Sepal Width (SW)')
plt.show()
```

```python
plt.hist(df['PL'], bins=10, color='orange', edgecolor='black')
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length (PL)')
plt.ylabel('Count')
plt.show()
```
