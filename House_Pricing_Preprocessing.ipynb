## House Price Dataset – Data Preprocessing

### Import Libraries and Load Dataset

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv(r"C:\Users\MidhunWork\DSA 2026 files\House_Pricing.csv")
df.head()
```

### Dataset Overview

```python
df.info()
df.describe()
```

### Duplicate Removal

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
duplicate_columns = []
columns = df.columns

for i in range(len(columns)):
    for j in range(i + 1, len(columns)):
        if df[columns[i]].equals(df[columns[j]]):
            duplicate_columns.append(columns[j])

print("Duplicate columns:", duplicate_columns)

df = df.drop(columns=duplicate_columns)

print("Dataset shape after duplicate removal:", df.shape)
```

### Handling Missing Values

```python
print(df.isnull().sum())
```

```python
df = df.dropna(subset=["Sale Price"])
```

```python
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_columns = df.select_dtypes(include=["object"]).columns.tolist()
```

```python
for column in numerical_columns:
    if column != "Sale Price":
        df[column] = df[column].fillna(df[column].median())
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print(df.isnull().sum())
```

### Scaling Numerical Variables

```python
df["Date House was Sold"] = pd.to_datetime(
    df["Date House was Sold"], format="%d %B %Y"
)
```

```python
scale_columns = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
scale_columns = [
    column for column in scale_columns
    if column not in ["Sale Price", "ID"]
]
```

```python
scaler = MinMaxScaler()
df[scale_columns] = scaler.fit_transform(df[scale_columns])

print("Scaled numerical columns:")
print(scale_columns)

df[scale_columns].head()
```

### Encoding Categorical Variables


```python
visit_mapping = { "None": 0, "Once": 1, "Twice": 2, "Thrice": 3, "Four": 4}
condition_mapping = { "Bad": 0, "Okay": 1, "Fair": 2, "Good": 3, "Excellent": 4}

df["No of Times Visited"] = df["No of Times Visited"].map(visit_mapping)
df["Condition of the House"] = df["Condition of the House"].map(condition_mapping)
```

```python
df = pd.get_dummies(df, columns=["Waterfront View"], drop_first=True, dtype=int)
df.head()
```

### Outlier Detection and Removal (IQR Method)

```python
outlier_columns = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
outlier_columns = [
    column for column in outlier_columns
    if column not in ["ID", "Zipcode", "Latitude", "Longitude"]
]
```

```python
outlier_rows = pd.Series(False, index=df.index)

for column in outlier_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outlier_rows = outlier_rows | (
        (df[column] < lower_limit) | (df[column] > upper_limit)
    )
```

```python
print("Number of rows identified as outliers:", outlier_rows.sum())
df = df[~outlier_rows].copy()
print("Shape after outlier removal:", df.shape)
```

### Final Dataset

```python
print("Final dataset shape:", df.shape)
print("\nMissing values in final dataset:")
print(df.isnull().sum())
df.head()
```

```python
df.to_csv("House_Pricing_Preprocessed.csv", index=False)
print("Preprocessed dataset saved successfully.")
```
