```python
rev=0
for order in orders:
    total_ord =order["quantity"]*order["price_per_item"]
    rev= rev+total_ord
print("Total revenue = ", rev)
```

```python
li=[]
for order in orders:
    total_ord= order["quantity"]*order["price_per_item"]
    if total_ord>2000:
        li.append(order)
print("Orders above Rs.2000 are: \n")

for order in li:
    print(order)
```

```python
item_quantity = {}
for order in orders:
    item =order["item"]
    if item in item_quantity:
        item_quantity[item] =item_quantity[item]+order["quantity"]
    else:
        item_quantity[item] =order["quantity"]
highest_item = ""
highest_quantity = 0
for item in item_quantity:
    if item_quantity[item] > highest_quantity:
        highest_quantity = item_quantity[item]
        highest_item = item
print(f"Highest quantity sold item is {highest_item} ,with quantity of: {highest_quantity}.")
```

```python
def apply_discount(orders, discount_rate):
    new_orders = []
    for order in orders:
        new_order = order.copy()
        discount = new_order["price_per_item"] * discount_rate
        new_order["price_per_item"] = new_order["price_per_item"] - discount
        new_orders.append(new_order)
    return new_orders
discounted_orders = apply_discount(orders, 0.10)
print("\nDiscounted orders\n")
for order in discounted_orders:
    print(order)
print("\nOriginal orders")
for order in orders:
    print(order)
```

```python
class OrderIterator:
    def __init__(self, orders):
        self.orders = orders
        self.position = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.position < len(self.orders):
            current_order = self.orders[self.position]
            self.position = self.position + 1
            return current_order
        else:
            raise StopIteration
print("Orders using iterator:")
obj = OrderIterator(orders)
for order in obj:
    print(order)
```

```python
customer_names = []
for order in orders:
    if order["customer"] not in customer_names:
        customer_names.append(order["customer"])
customer_orders = {
    customer: [order for order in orders if order["customer"] == customer]
    for customer in customer_names
}
print("Orders grouped by customer\n")
for customer in customer_orders:
    print(customer)
    for order in customer_orders[customer]:
        print(order)
```
