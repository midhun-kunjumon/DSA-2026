# Bank Account Manager


```python
bank_accs = { 101:['Abhijith A',70500], 102:['Babu K',3501], 103:['Denin John',102030], 104:['Asfar T',51320]
}
def deposit(acc_no):
    dep_amnt=int(input("Enter amount to deposit: "))
    bank_accs[acc_no][1]=bank_accs[acc_no][1]+dep_amnt
    print(f"\n Transaction successful")
    print(f"account holder: {bank_accs[acc_no][0]}")
    print(f"current balance: {bank_accs[acc_no][1]}")
def withdraw(acc_no):
    wd_amnt=int(input("Enter amount to withdraw: "))
    current_balance=bank_accs[acc_no][1]
    if wd_amnt>current_balance:
        print(f"\n Transaction failed: Insufficient balance")
        print(f"available balance:{current_balance}")
    else:
        bank_accs[acc_no][1]=bank_accs[acc_no][1]-wd_amnt
        print(f"\n Transaction successful")
        print(f"account holder: {bank_accs[acc_no][0]}")
        print(f"current balance: {bank_accs[acc_no][1]}")
def balance(acc_no):
    print(f"\n Account details:")
    print(f"account holder: {bank_accs[acc_no][0]}")
    print(f"current balance: {bank_accs[acc_no][1]}")
acc_no=int(input("Enter the bank account n.o.:"))
if acc_no in bank_accs:
    name=bank_accs[acc_no][0]
    con=input(f"Are you {name}? (Y/N): ").strip().upper()
    if con=='Y':
        op=int(input("Enter operation number \n Operations: 1 = Deposit/ 2 = Withdrawal/ 3 = Balance enquiry: "))
        if op==1:
            deposit(acc_no)
        elif op==2:
            withdraw(acc_no)
        elif op==3:
            balance(acc_no)
        else:
            print("Invalid operation selected")
    else:
        print("Identity verification failed")
else:
    print("Incorrect bank acc n.o.")
```

# Weather Data Anayzer

```python
def wthr_anlyzr(ptn,cc,hum,temp):
  if (ptn>0 and cc>=75 and hum>=70 and temp<=5):
    wthr="Snowy"
  elif (ptn>0 and cc>=75 and hum>=70 and 15>temp>5):
    wthr="Rainy"
  elif (ptn==0 and 71<cc<=100 and hum>=60 and 25>temp>=10):
    wthr="Cloudy"
  elif (ptn==0 and 21<cc<=70 and 40>hum>=65 and 35>temp>=15):
    wthr="Partly Cloudy"
  elif (ptn==0 and 0<=cc<=20 and 40<=hum and temp>=25):
    wthr="Sunny"
  else:
    wthr="Undefined Pattern can't predict the weather."
  return wthr
print("\n To analyze the weather enter the following parameters.")
temp=int(input("\n Temprature(°C): "))
hum=int(input("\n Humidity(%): "))
ptn=int(input("\n Precipitation(mm): "))
cc=int(input("\n Cloud Coverage(%): "))
print(f"\n Weather Condition: {wthr_anlyzr(ptn,cc,hum,temp)}")
```
